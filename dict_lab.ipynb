# Python Dictionaries

- Data Collection. Keyed Collections
- Properties | Key: Value pairs
- Any data type

```python
student_info = {
    "name": "Dylan",
    "age": 30,
    "courses": ["Business Intelligence"],
    "is_enrolled": True
}

# Lists - list of nouns (list of animals, list of names, list of groceries)
# Dict - single noun and all of the properties describe the noun

print(type(student_info))
```

```python
print(student_info['name'])
print(student_info.get('name'))
```

```python
#Modifying Dict Info

# updating exisiting value
student_info['age'] = 33

#creating new key:value pairs
student_info['grade'] = 'A'

print(student_info)
```

```python
# call a specific property you can use the data type
student_info['courses'].append('Data Science')
student_info['courses'].extend(['BI', 'Data Engineering'])

#you can use all methods on key val pairs
student_info['name'] = student_info['name'].upper()

print(student_info)
```

```python
print(student_info['courses'][0:2])
```

```python
#Removing properties from a dict
#del student_info['grade']

#Pop to delete from a dict (saves deleted value) gets returned by method
removed_value = student_info.pop("is_enrolled")

print(f'Removed Value: {removed_value}')

print(student_info)
```

```python
# Common Dict Methods
print(student_info.keys())
print(student_info.values())
```

```python
print(student_info.items()) # returns tuples of the key val pairs, immutable (can't be reordered)
```

```python
for pair in student_info.items():
    print(pair)
```
